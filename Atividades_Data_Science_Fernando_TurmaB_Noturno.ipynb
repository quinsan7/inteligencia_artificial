4. Gere um array de 100 numeros inteiros aleatorios e encontre a mediana e
a moda dos valores.

```python
import numpy as np
array = np.random.randint(1, 101, 100)
mediana = np.median(array)

contagem = {}
for numero in array:
  if numero in contagem:
    contagem[numero] += 1
  else:
    contagem[numero] = 1

moda = 0
maior_contagem = 0
for numero in contagem:
  if contagem[numero] > maior_contagem:
    maior_contagem = contagem[numero]
    moda = numero

print(mediana)
print(moda)
```

7. Crie dois arrays de numeros aleatorios e encontre a correlacao entre eles.

```python
import numpy as np

array1 = np.random.rand(20)
array2 = np.random.rand(20)

media1 = np.mean(array1)
media2 = np.mean(array2)
desvio1 = np.std(array1)
desvio2 = np.std(array2)

covariancia = np.mean((array1 - media1) * (array2 - media2))
correlacao = covariancia / (desvio1 * desvio2)
print("Correlação:", correlacao)
```

19. Crie dois arrays de numeros inteiros aleatorios de tamanho 20 e calcule o
angulo entre eles (produto escalar normalizado).

```python
import numpy as np
import math

array1 = np.random.randint(1, 10, 20)
array2 = np.random.randint(1, 10, 20)

produto_escalar = np.dot(array1, array2)
norma1 = math.sqrt(np.sum(array1 ** 2))
norma2 = math.sqrt(np.sum(array2 ** 2))

cos_angulo = produto_escalar / (norma1 * norma2)
angulo = math.acos(cos_angulo)
print(angulo)
```

1. Calcule a integral da funcao f(x) = x
2
sin(x) no intervalo de 0 a π.

```python
from scipy import integrate
import numpy as np

def f(x):
    return x**2 * np.sin(x)
resultado, erro = integrate.quad(f, 0, np.pi)

print(resultado)
```

2. Resolva a equacao diferencial dy
dx = x − y, com y(0) = 1, no intervalo de 0
a 10.

```python
from scipy.integrate import solve_ivp
import numpy as np

def eq(t, y):
    return t - y
solucao = solve_ivp(eq, [0, 10], [1], t_eval=np.linspace(0, 10, 50))

print(solucao.y[0])
```

10. Calcule a integral dupla da funcao f(x, y) = xy + x
2 no intervalo [0, 2]
para x e [0, 1] para y

```python
from scipy import integrate

def f_interna(y):
    def f(x):
        return x*y + x**2
    resultado, erro = integrate.quad(f, 0, 2)
    return resultado

resultado_total, erro = integrate.quad(f_interna, 0, 1)

print(resultado_total)
```

15. Calcule a transformada de Fourier inversa de um sinal complexo gerado
aleatoriamente.

```python
from scipy.fft import fft
import numpy as np

parte_real = np.random.rand(50)
parte_imaginaria = np.random.rand(50)
sinal = parte_real + 1j*parte_imaginaria

transformado = fft(sinal)

n = len(transformado)
inverso = fft(transformado) / n
inverso = np.conj(inverso)

print(inverso)
```

8. Plote um grafico de linha com intervalos de confianca (sombramento) para
a funcao y = sin(x).

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)
y_superior = y + 0.2
y_inferior = y - 0.2

plt.plot(x, y)
plt.fill_between(x, y_inferior, y_superior, alpha=0.3)
plt.title("Função Seno com Intervalo de Confiança")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")

plt.show()
```

10. Plote tres graficos de linha em um unico grafico, cada um com uma cor e
estilo de linha diferentes.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

plt.plot(x, y1, color="red", label="Seno")
plt.plot(x, y2, color="blue", label="Cosseno")
plt.plot(x, y3, color="green", label="Seno x Cosseno")
plt.title("Três Linhas")
plt.legend()

plt.show()
```

11. Crie um grafico de barras empilhadas para tres categorias com tres grupos
de valores.


```python
import matplotlib.pyplot as plt

categorias = ["A", "B", "C"]
grupo1 = [3, 5, 2]
grupo2 = [2, 3, 4]
grupo3 = [1, 2, 3]

soma_grupo1_2 = []

for i in range(3):
    soma_grupo1_2.append(grupo1[i] + grupo2[i])

plt.bar(categorias, grupo1, label="Grupo 1")
plt.bar(categorias, grupo2, bottom=grupo1, label="Grupo 2")
plt.bar(categorias, grupo3, bottom=soma_grupo1_2, label="Grupo 3")
plt.title("Barras Empilhadas")
plt.legend()

plt.show()
```

18. Crie um grafico de barras com valores positivos e negativos e use diferentes
cores para cada tipo de valor.

```python
import matplotlib.pyplot as plt
categorias = ["A", "B", "C", "D"]
valores = [5, -3, 7, -2]
cores = []
for v in valores:
    if v > 0:
        cores.append("green")
    else:
        cores.append("red")
plt.bar(categorias, valores, color=cores)
plt.title("Valores Positivos e Negativos")

plt.show()
```

20. Adicione uma grade com espacamento personalizado a um grafico de linha
para y = x
2


```python
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator
import numpy as np

x = np.linspace(-10, 10, 100)
y = x**2

fig, eixo = plt.subplots()
eixo.plot(x, y)
eixo.set_title("Função Quadrática com Grade Personalizada")

eixo.xaxis.set_major_locator(MultipleLocator(2))
eixo.yaxis.set_major_locator(MultipleLocator(10))
eixo.grid(True)

plt.show()
```
