21.Busca Binária

* Implemente a busca binária para encontrar um elemento em uma
lista ordenada.

Explicação: Marco o começo e o fim da lista. Toda vez eu olho o número que tá bem no meio desse intervalo. Se for o número que eu quero, acabou. Se for menor que o que eu busco, sei que ele tá na metade de cima, então movo o início. Se for maior, movo o fim. Vou apertando o intervalo até achar ou não sobrar mais nada.

```python
def bb(lista,buscar):
  inicio = 0
  fim = len(lista) -1

  while inicio <= fim:
    meio = (inicio + fim) //2

    if lista[meio] == buscar:
      return meio
    elif lista[meio] < buscar:
      inicio = meio +1
    else:
      fim = meio - 1

  return -1

lista = [1,2,3,4,5,6,7,8,9]
buscar = 7
resultado = bb(lista,buscar)
print(resultado)
```

22. Criptografia de Cifra de César

*   Crie uma função para criptografar uma mensagem usando a cifra de
César.

Explicação: Fiz duas strings com o alfabeto, uma minúscula e uma maiúscula. Pra cada letra da mensagem, procuro ela dentro do alfabeto certo usando um for que vai comparando posição por posição, guardo o número da posição, somo o deslocamento e pego a letra que sobrou nesse novo lugar. Repeti a lógica pras duas versões do alfabeto porque não sabia um jeito mais direto de juntar as duas.

```python
alfabeto_minusculo = "abcdefghijklmnopqrstuvwxyz"
alfabeto_maiusculo = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

def cifra_cesar(mensagem, deslocamento):
  resultado = ""
  for letra in mensagem:
    if letra in alfabeto_minusculo:
      posicao = 0
      for i in range(len(alfabeto_minusculo)):
        if alfabeto_minusculo[i] == letra:
          posicao = i
      nova_posicao = (posicao + deslocamento) % 26
      resultado = resultado + alfabeto_minusculo[nova_posicao]
    elif letra in alfabeto_maiusculo:
      posicao = 0
      for i in range(len(alfabeto_maiusculo)):
        if alfabeto_maiusculo[i] == letra:
          posicao = i
      nova_posicao = (posicao + deslocamento) % 26
      resultado = resultado + alfabeto_maiusculo[nova_posicao]
    else:
      resultado = resultado + letra
  return resultado

texto = input("Digite uma palavra: ")
deslocamento = 3
crip = cifra_cesar(texto, deslocamento)
print(texto)
print(crip)
```

23. Descriptografar Cifra de César

*  Crie uma função para descriptografar uma mensagem cifrada com a
cifra de César.

Explicação: Usei a mesma ideia da cifra, só que voltando as posições em vez de avançando. Procuro a letra no alfabeto com um for, pego a posição dela, subtraio o deslocamento e uso o % 26 pra não dar posição negativa.

```python
alfabeto = "abcdefghijklmnopqrstuvwxyz"

def descriptografar(mensagem, deslocamento):
  resultado = ""
  for letra in mensagem.lower():
    if letra in alfabeto:
      posicao = 0
      for i in range(len(alfabeto)):
        if alfabeto[i] == letra:
          posicao = i
      nova_posicao = (posicao - deslocamento) % 26
      resultado = resultado + alfabeto[nova_posicao]
    else:
      resultado = resultado + letra
  return resultado

texto = input("Digite a mensagem cifrada: ")
deslocamento = 3
descrip = descriptografar(texto, deslocamento)
print(texto)
print(descrip)
```

24.Validação de CPF

*   Implemente uma função para validar um CPF brasileiro.

Explicação: Tirei o uso de regex porque achei complicado de mexer, então fiz na mão: passo letra por letra do CPF e só guardo o que for número, jogando fora ponto e traço. Depois confiro se sobraram 11 dígitos e se não são todos iguais, comparando cada um com o primeiro. A parte dos dígitos verificadores eu peguei pesquisando como funciona o cálculo oficial, multiplicando cada número por um peso e tirando o resto da divisão por 11 — fiz duas vezes porque são dois dígitos pra conferir.

```python
def validar(cpf):
  cpf_limpo = ""
  for caractere in cpf:
    if caractere.isdigit():
      cpf_limpo = cpf_limpo + caractere
  cpf = cpf_limpo

  if len(cpf) != 11:
    return False

  igual = True
  for i in range(len(cpf)):
    if cpf[i] != cpf[0]:
      igual = False
  if igual:
    return False

  soma = 0
  for i in range(9):
    soma = soma + int(cpf[i]) * (10 - i)
  resto = (soma * 10) % 11
  if resto == 10:
    primeiro = 0
  else:
    primeiro = resto
  if primeiro != int(cpf[9]):
    return False

  soma = 0
  for i in range(10):
    soma = soma + int(cpf[i]) * (11 - i)
  resto = (soma * 10) % 11
  if resto == 10:
    segundo = 0
  else:
    segundo = resto
  if segundo != int(cpf[10]):
    return False

  return True

cpf_usuario = input("Digite se cpf: ")
if validar(cpf_usuario):
  resultado = "Valido"
else:
  resultado = "Invalido"
print(f"{cpf_usuario}: {resultado}")
```

26. Calculadora Simples

*  Crie uma calculadora que possa realizar adição, subtração,

27.Anagramas

* Verifique se duas palavras são anagramas (ou seja, se são formadas
pelas mesmas letras).

Explicação: Fui "riscando" cada letra da palavra1 de uma cópia da palavra2 conforme ia encontrando ela. Se sobrar alguma letra sem par no final, ou se uma letra não existir pra riscar, então não são anagramas.

```python
def engramas(palavra1, palavra2):
  palavra1 = palavra1.replace(" ", "").lower()
  palavra2 = palavra2.replace(" ", "").lower()

  if len(palavra1) != len(palavra2):
    return False

  letras_restantes = list(palavra2)
  for letra in palavra1:
    if letra in letras_restantes:
      letras_restantes.remove(letra)
    else:
      return False

  return len(letras_restantes) == 0

palavra3 = input("Digite uma palavra: ")
palavra4 = input("Digite uma palavra: ")

if engramas(palavra3, palavra4):
  print("Sao anagramas")
else:
  print("Não são anagramas")
```

30.Jogo da Forca

*   Desenvolva um jogo simples da forca onde o usuário deve adivinhar
uma palavra.

Explicação: Sorteio uma palavra e crio uma lista de traços do mesmo tamanho dela, uma posição pra cada letra. O jogo roda enquanto tiver tentativa e enquanto ainda tiver traço sobrando. Quando a letra bate, procuro todas as posições dela na palavra (porque pode repetir) e vou revelando.

```python
import random

palavras = ["carro","moto","aviao","onibus","caminhao","pes"]

adivinhar = random.choice(palavras)
letras_certas = ["_"] * len(adivinhar)
letras_erradas = []

tentativas = 6

print("Jogo da Forca")
print("".join(letras_certas))

while tentativas >0 and "_" in letras_certas:
  letra = input("Digite uma letra: ").lower()

  if letra in letras_erradas:
    print("Essa letra ja foi")
    continue

  letras_erradas.append(letra)

  if letra in adivinhar:
    for i in range(len(adivinhar)):
      if adivinhar[i] == letra:
        letras_certas[i] = letra
    print("Essa letra esta presenta na palavra")
  else:
    tentativas -=1
    print(f"Essa letra não esta presenta na palavra, {tentativas}")

  print("".join(letras_certas))

if "_" not in letras_certas:
  print(f"Boa! Voçê acertou a palavra, {adivinhar}")
else:
  print(f"Que pena! Voçê errou, {adivinhar}")
```

31.Ordenação por Bolha

*  Implemente o algoritmo de ordenação por bolha (Bubble Sort).

Explicação: Usei uma variável trocou pra controlar um while, em vez de dois for aninhados. A ideia é: se em uma passada inteira eu não precisei trocar nada, é porque a lista já tá ordenada e posso parar — assim evito repetir passadas desnecessárias.

```python
def bolha(lista):
  trocou = True
  while trocou:
    trocou = False
    for j in range(len(lista) - 1):
      if lista[j] > lista[j + 1]:
        guardar = lista[j]
        lista[j] = lista[j + 1]
        lista[j + 1] = guardar
        trocou = True
  return lista

numeros = [21,52,854,51,94,163,987,612]
ordenar = bolha(numeros)
print(ordenar)
```

```python
colunas = ["1","2","3","4","5","6","7","8","9"]

def tabuleiro():
  print(colunas[0] + "|" + colunas[1] + "|" + colunas[2])
  print("--+--+--")
  print(colunas[3] + "|" + colunas[4] + "|" + colunas[5])
  print("--+--+--")
  print(colunas[6] + "|" + colunas[7] + "|" + colunas[8])


def ganhador():
  verificacao = [[0,1,2],[3,4,5],[6,7,8],
                [0,3,6],[1,4,7],[2,5,8],
                [0,4,8],[2,4,6]]

  for verificar in verificacao:
    a = verificar[0]
    b = verificar[1]
    c = verificar[2]

    if colunas[a] == colunas[b] == colunas[c]:
      return colunas[a]

def deu_velha():
  for posicao in colunas:
    if posicao != "X" and posicao != "O":
      return False
  return True

jogador = input("Voçê quer ser X ou O").upper()

while jogador != "X" and jogador != "O":
  jogador = input("Isso nao é x nem o").upper()

continua = 1

while continua:
  print(f"É a vez do {jogador}")
  posicao = input("escolha uma posicao de 1 a 9: ")
  indice = int(posicao) - 1

  if colunas[indice] == "X" or colunas[indice] == "O":
    print("local ocupado! escolha outra casa.")
    continue

  colunas[indice] = jogador
  tabuleiro()

  vencedor = ganhador()

  if vencedor:
    print(f"{vencedor} ganhou")
    continua = 0
  elif deu_velha():
    print("Deu velha")
    continua = 0
  else:
    if jogador == "X":
      jogador = "O"
    else:
      jogador = "X"
```

33.Verificação de Anagramas

*   Verifique se duas palavras fornecidas são anagramas uma da outra.

Explicação: Em vez de usar sorted, contei quantas vezes cada letra aparece em cada palavra usando dois dicionários separados. Se as duas contagens ficarem iguais, é porque as palavras têm exatamente as mesmas letras na mesma quantidade.

```python
def anagramas(palavra1, palavra2):
  palavra1 = palavra1.replace(" ", "").lower()
  palavra2 = palavra2.replace(" ", "").lower()

  if len(palavra1) != len(palavra2):
    return False

  contagem1 = {}
  contagem2 = {}

  for letra in palavra1:
    if letra in contagem1:
      contagem1[letra] += 1
    else:
      contagem1[letra] = 1

  for letra in palavra2:
    if letra in contagem2:
      contagem2[letra] += 1
    else:
      contagem2[letra] = 1

  return contagem1 == contagem2

palavra3 = input("Digite uma palavra: ")
palavra4 = input("Digite uma palavra: ")

if anagramas(palavra3, palavra4):
  print("Sao anagramas")
else:
  print("Não são anagramas")
```

37. Funções Recursivas

*   Implemente uma função recursiva para calcular o fatorial de um
número.

Explicação: A função chama ela mesma dentro dela, cada vez com um número menor. Isso só para quando chega em 0 ou 1, que aí ela devolve 1 direto sem chamar de novo. Multiplicando tudo isso na volta dá o fatorial.

```python
def fatorial(n):
  if n == 0 or n == 1:
    return 1
  return n * fatorial(n - 1)

numero = int(input("Digite um numero: "))
resultado = fatorial(numero)

print(resultado)
```

```python
n = int(input("Digite o valor de N: "))

marcado = []
for i in range(n + 1):
  marcado.append(True)

marcado[0] = False
if n >= 1:
  marcado[1] = False

numero = 2
while numero <= n:
  if marcado[numero] == True:
    print(numero)
    contador = numero + numero
    while contador <= n:
      marcado[contador] = False
      contador = contador + numero
  numero = numero + 1
```

46. Calcular Média e Desvio Padrão

*  Calcule a média e o desvio padrão de uma lista de números.

Explicação: A média é só soma dividida pela quantidade. Pro desvio padrão, pego a diferença de cada número pra média, elevo ao quadrado (senão os negativos cancelam os positivos) e somo tudo. No final tiro a raiz dividindo por 0.5.

```python
numeros = [10, 20, 30, 40, 50]

media = sum(numeros) / len(numeros)

soma_diferencas = 0
for numero in numeros:
  soma_diferencas += (numero - media) ** 2

desvio_padrao = (soma_diferencas / len(numeros)) ** 0.5

print(f"Média: {media}")
print(f"Desvio padrão: {desvio_padrao}")
```

48.Ordenação por Inserção

*   Implemente o algoritmo de ordenação por inserção (Insertion Sort).

Explicação: Pego cada número a partir do segundo e guardo numa variável. Fico comparando ele com os números antes dele: se o de trás for maior, empurro ele uma casa pra frente e continuo olhando pra trás; se não for maior, uso uma variável pra parar o loop. No fim coloco o número guardado no lugar certo.

```python
def insercao(lista):
  for i in range(1, len(lista)):
    atual = lista[i]
    j = i - 1
    parar = False
    while j >= 0 and parar == False:
      if lista[j] > atual:
        lista[j + 1] = lista[j]
        j = j - 1
      else:
        parar = True
    lista[j + 1] = atual
  return lista

numeros = [21, 52, 854, 51, 94, 163, 987, 612]
ordenar = insercao(numeros)
print(ordenar)
```
